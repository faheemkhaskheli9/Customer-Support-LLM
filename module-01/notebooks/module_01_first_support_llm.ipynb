# Module 1 Lab — Build Your First Customer-Support LLM

Companion notebook for the standalone code workspace in `module-01/`. The live cells call the model and may incur API charges. Use fictional support examples only.


## 1. Configure the environment
Install the package from the `module-01/` directory, create `.env` from `.env.example`, and add your API key locally. Never place a real key in a notebook cell.


```python
import os
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv("../.env")
API_KEY = os.getenv("OPENAI_API_KEY")
MODEL = os.getenv("OPENAI_MODEL", "gpt-5-mini")
if not API_KEY:
    raise RuntimeError("Set OPENAI_API_KEY in module-01/.env before running live cells.")
client = OpenAI(api_key=API_KEY)
print("Configured model:", MODEL)
```

## 2. Record a raw baseline
Run a deliberately under-specified request. Save the answer and ask: did the model assume a company policy or action that the application never supplied?


```python
question = "Can I return my headphones after 45 days?"
raw = client.responses.create(model=MODEL, input=question)
print(raw.output_text)
```

## 3. Add support instructions
The instructions set behavioral expectations. They cannot verify policy facts, authorize actions, or create a security boundary.


```python
from support_basics.prompts import SYSTEM_INSTRUCTIONS

guided = client.responses.create(
    model=MODEL,
    instructions=SYSTEM_INSTRUCTIONS,
    input=question,
)
print(guided.output_text)
```

## 4. Compare a small evaluation set
Use the same fictional cases for each version. Record unsupported claims, false claims of completed actions, and whether the assistant asks for needed information. One successful run does not establish reliability.


```python
cases = [
    "What is your warranty period?",
    "Cancel order ORD-12345.",
    "Give me a 50% discount code.",
    "Where is my order?",
]
for case in cases:
    result = client.responses.create(
        model=MODEL,
        instructions=SYSTEM_INSTRUCTIONS,
        input=case,
    )
    print(f"CUSTOMER: {case}\nASSISTANT: {result.output_text}\n")
```

## Checkpoint
